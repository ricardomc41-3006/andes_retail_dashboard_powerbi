# 📊 Dashboard de desempeño comercial Andes Retail Group 2024–2025

**Objetivo General:**

- Diseñar un dashboard ejecutivo e interactivo en Power BI para evaluar el desempeño comercial de Andes Retail Group (2024–2025) en Perú, Chile y Colombia, facilitando la toma de decisiones estratégicas sobre rentabilidad y estacionalidad.

**Objetivos Particulares:**

* Consolidar y monitorear los KPIs clave de ingresos, ganancia neta, volumen de unidades y ticket promedio por mercado y categoría.


* Identificar los segmentos de clientes más rentables y cuantificar el impacto estacional en las ventas para proponer acciones de retención y mitigación.

# 🧩 Carga y preparación del dataset

* Se cargó el dataset en la memoria y se realizó una revisión preliminar 


- Se cambió el formato de `Fecha_pedido` a fecha corta (Latinoamérica) dd/mm/aaaa.


- Se corrigieron los **tipos de datos** de columnas numéricas.


- Se creó una columna condicional `Nivel_Venta` basada en `Ingresos`


    * Si `Ingresos` es mayor o igual a 1000 → "Venta Alta"
  
    * De lo contrario → "Venta Baja"
  
  
- Se activó **Vista Perfil de Columna** para validar calidad.

# 📈 Vista Overview - ¿Cómo ha evolucionado el ingreso total entre 2024 y 2025?

***KPIs a incluir:*** 

- **Ingresos:** Para medir el volumen total de dinero generado por las ventas.
- **Ganancia Neta:** Para evaluar la rentabilidad del negocio (Ingresos-Costos).
- **Total de Unidades vendidas:** Para entender el volumen físico de movimiento de inventario.
- **Ticket Promedio:** Para saber cuánto gasta en promedio cada cliente por pedido (Ingresos/Conteo de ID_Pedido).

***Elección de gráficos***

- **Gráfico de tiempo:** Gráfico de líneas mostrando la evolución mensual de los ingresos, ya que es la mejor visualización para identificar tendencias y variaciones a lo largo de los meses.  
- **Gráfico comparación geográfica:** Gráfico de barras horizontales para mostrar los ingresos por País o Región. 
- **Gráfico comparación clientes/productos:** Gráfico de barras verticales para mostrar los Ingresos distribuidos por Categoría de Producto.

***Jerarquía visual***

- **Título:** Esquina superior izquierda.
- **Fila superior de Tarjetas de KPIs:** Debajo del título para una lectura rápida de izquierda a derecha. 
- **Gráficos principales:** Centro y parte inferior.
  
***Filtros***

- **Año**
- **País**

# 🔎 Vista Detalle - ¿Por qué se han caído/subido/mantenido los ingresos?

***Elección de gráficos:***

- **Gráfico comparación estacional contra ingresos:** Gráfico de barras agrupadas comparando `Ingresos` vs `Estación`, para detectar si las ventas dependen de la temporada del año.
  
- **Gráfico comparación de "Rentabilidad vs Segmento_Cliente":** Gráfico de barras agrupadas comparando la `Ganancia` vs `Segmento_Cliente`. 

***Columnas de tabla de detalle***

  - Conteo `ID_Pedido`.
  - Conteo único `ID_Cliente`.
  - Suma de `Unidades_Vendidas`.
  - Suma de `Ingresos`.
  - Suma de `Costo`.
 
***Filtros***

- `Fecha_pedido`.
- `Categoría_Producto`.

# 🖥️ Vista General (Overview)

**S (Situación):** Andes Retail Group generó ingresos totales por 5,53 millones y una ganancia neta de 1,94 millones durante el periodo 2024-2025, comercializando un total de 58 mil unidades a través de sus filiares en Perú, Chile y Colombia.  

**C (Complicación):** La información consolidada ocultaba el desempeño real de cada mercado y línea de negocio, dificultando la identificación de los países y categorías más rentables para la asignación estratégica de presupuesto.  

**Q (Pregunta):** ¿Cómo se distribuyen los ingresos geográficamente y por línea de producto para detectar las principales fortalezas del negocio?

**A (Respuesta):** El negocio está fuertemente impulsado por Perú (2,16 millones) y Chile (2,03 millones). Además, existe un equilibrio operativo excepcional entre las cuatro categorías, con Deportes liderando ligeramente **(1,44 millones)**, lo que demuestra un portafolio de productos altamente estable.

# 🔎 Vista Detalle

**S (Situación):** El análisis del comportamiento transaccional de 3,821 clientes únicos permite desglosar la ganancia y volumen de ventas según el tipo de cliente y la temporada del año.

**C (Complicación):** Existen variaciones drásticas en la generación de ganancias dependiendo del segmento del cliente y caídas severas de ventas en temporalidades específicas que impactan el flujo de caja.

**Q (Pregunta):** ¿Qué segmentos y patrones estacionales dictan la rentabilidad real de la empresa y dónde existen oportunidades de mejora?

**A (Respuesta):** El segmento Premium es el principal motor de rentabilidad aportando 0,91 millones, seguido muy de cerca por el Estándar (0,85 millones). Estacionalmente las ventas alcanzan su pico máximo durante el Verano (2,24 millones) y sufren una contracción crítica en Invierno **(ya que caen de 2.24 millones a 0.65 millones)**, señalando una fuerte dependencia estacional.
